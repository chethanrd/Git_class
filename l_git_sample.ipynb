```python
calls=int(input("Enter Number of call from the Customer"))
if calls<=100:
    charge1=calls*2
    print(f"Th Telophone number which customer make {calls} is :", charge1)

elif calls>100 and calls<=150:
    charge2=(calls*2)+(calls*0.6)
    print(f"Th Telophone number which customer make {calls} is :", charge2)

elif calls>150 and calls<=200:
    charge3=(calls*2)+(calls*0.5)
    print(f"Th Telophone number which customer make {calls} is :", charge3)

else:
    charge4=(calls*2)+(calls*0.4)
    print(f"Th Telophone number which customer make {calls} is :", charge4)
```

```python
calls=int(input("Enter the number of calls which Customer make in a month"))
total_bill=0.0
if calls<=100:
    total_bill=200
    print(f"Total billl of which customer make {calls} calls in a month is :", total_bill)

elif calls<=150:
    total_bill=200+(0.6 *(calls-100))
    print(f"Total billl of which customer make {calls} calls in a month is :", total_bill)

elif calls<=200:
    total_bill=200+ (0.6*50) +(0.5 *(calls-150))
    print(f"Total billl of which customer make {calls} calls in a month is :", total_bill)

else:
    total_bill=200+(0.6*50)+(0.5*50)+(0.4 *(calls-200))
    print(f"Total billl of which customer make {calls} calls in a month is :", total_bill)
```
