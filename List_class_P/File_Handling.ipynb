```python
def is_even(num):
    return True if num%2==0 else False
```

```python
def my_filter(func,seq):
    out_lst=[]
    for ele in seq:
        if func(ele)==True:
            out_lst.append(ele)
    return out_lst
```

```python
my_filter(is_even,lst)
```
