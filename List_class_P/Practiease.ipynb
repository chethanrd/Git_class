```python
def split_str(my_str):
    out_lst=[]
    word=''
    for ind in range(len(my_str)):
        if my_str[ind]==' ':
            out_lst+=[word]
            word=''
        else:
            word+=my_str[ind]
    else:
        out_lst+=[word]
    print(out_lst)


```

```python
split_str('peter piper picked a peck of pickled peppers')
```
