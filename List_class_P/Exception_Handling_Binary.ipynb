```python
# try, exept, else, finally, raise 
```

```python
# try - we write the piece of code which is suppose to raise the error.
#except - The code which you want to execute if error is raised. 
# else - The code which you want to execute if error is not raised. 
# finally - The code which you want to execute irrespective of the errors.  
# raise - used to raise the custom errors. 
```

```python
a=int(input('Enter a Number: '))
b=int(input('Enter a Number: '))
try:
    res=a/b
except:
    print('Division by 0 is not allowed')
else:
    print("Else block:",res)
finally:
    print(a)
    print(b)
```

```python
# 1. Each try block should have at least one except block
# 2. one try block may have a multiple  except block
# 3. one except block can not have amultiple try block
# 4. The order of try except should be try -> except ->else ->finally, where else and finally are optional
# 5. Nested try , you should have nested except block as well. 
# 6. child exception should be followed by parent exception. 
```

```python
# try with multiple except block.
try:
    a=int(input('Enter a Number: '))
    b=int(input('Enter a Number: '))
    res=a/b

except ArithmeticError as el:
    print('Arithemeticerror')
    print(el)
except ZeroDivisionError as e2:
    print('ZeroDivsionError')
    print(e2)
except ValueError as e3:
    print('ValueError')
    print(e3)
except Exception as e:
    print('Exception Block')
    print(e)
else:
    print("Else block:",res)
finally:
    print(a)
    print(b)
```

```python
# nested try-except block
try:
    a=int(input('Enter a Number: '))
    b=int(input('Enter a Number: '))
    try:
        res=a/b
    except:
        print("Division by zero is not allowed")
except:
    print('Some error')
else:
    print("else")
```

```python
# Userdefined Exceptions

class MyException(Exception):
    def __init__(self,msg):
        self.msg=msg
```

```python
def amount_withdraw(amt):
    if amt > 49999:
        raise MyException('Need Pan for this amount')
    else:
        print('Amount Withdrawn')
        print('Please collect you are cash')

def pan_valid(pan_no):
    if len(pan_no)==10:
        print('pan is valid')
        print('Request is processed')
    else:
        raise MyException('Invalid Pan')
```

```python
try:
    amount_withdraw(50000)
except MyException as e :
    print(e)

    pan_no=input('Enter you are pan no:')
    try:
        pan_valid(pan_no)
    except MyException as e :
        print(e)
```
