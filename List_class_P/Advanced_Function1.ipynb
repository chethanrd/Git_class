```python
nums = [10, 15, 17, 4, 23, 8, 29, 33]
def prime_number(numbers):
    if numbers<2:
        return False
    else:
        for i in range(2,int(math.sqrt(numbers))+1):
            if numbers%i==0:
                return False
        return True
  
```

```python
prime_numbers_result=list(filter(prime_number,nums))
print(prime_numbers_result)
```

```python
emails = ["abc@gmail.com", "invalidemail", "xyz@yahoo", "test@domain.com", "noatsign.com","cheth.an@gmail.com","xyz@gmail.com"]
def is_valid(email):
    if email.count('@')!=1:
        return False
    username,domain=email.split('@')
    
    if domain.count('.')!=1:
        return False

    if domain.startswith('.') and domain.endswith('.'):
        return False

    if not username or not domain:
        return False

    return True


email_result=list(filter(is_valid,emails))
print(email_result)
```
