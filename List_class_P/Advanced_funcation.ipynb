```python
def subtrac_num(num1,num2):
    num1=int(input("Enter First Number"))
    num2=int(input("Enter Second Number"))
    total=num1-num2
    return total
    
print(subtrac_num(20,10))
```

```python
def biger_num():
    num1=int(input("Enter the first number to compare"))
    num2=int(input("Enter the second number to compare"))
    if num1>num2:
        return num1
    else:
        return num2

```

```python
print(biger_num())
```

```python
def tempertue_conversion():
    c_temp=float(input("Enter a Temperature to convert"))
    formula=((c_temp*9)/5)+32
    convert_str=str(round(formula,2))
    print("Convertion of temperature from Celsius to Fahrenheit scale is:",convert_str)
```

```python
tempertue_conversion()
```

```python
def conv_BMI(weight,height):
    bmi=(weight)/(height*height)
    final_bmi=str(round(bmi,1))
    return final_bmi
    
```

```python
conv_BMI(weight=63,height=1.7)
```
