```python
def tempertue_conversion():
    c_temp=float(input("Enter a Temperature to convert"))
    formula=((c_temp*9)/5)+32
    convert_str=str(round(formula,2))
    print("Convertion of temperature from Celsius to Fahrenheit scale is:",convert_str)
```

```python
def add_lastnumber():
    num_list=eval(input("Enter list of numbers"))
    print(num_list)
    lst_two_num=num_list[-1]
    f_num=int(str(lst_two_num)[-2])
    s_num=int(str(lst_two_num)[-1])
    sum=(f_num+s_num)

    return sum
        

#print(sum)
```

```python
add_lastnumber()
```

```python
def add_lastnumber1():
    num_list=eval(input("Enter the list of numbers in list form"))
    print(num_list)
    lst_num=num_list[-1]
    ltwo_num=lst_num%100
    f_num=ltwo_num//10
    s_num=ltwo_num%10
    sum_twonum=f_num+s_num
    return sum_twonum
```
